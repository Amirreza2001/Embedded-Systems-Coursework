## **Merge Sort Algorithm**

Merge Sort is an efficient, comparison-based sorting algorithm that follows the "Divide and Conquer" paradigm. It repeatedly divides the input list into two halves until it reaches sub-lists containing a single element (which are inherently sorted). Then, these single-element sub-lists are merged back together in sorted order, pair by pair. This merging process continues until all sub-lists are combined into a single, fully sorted list.

### **How it Works:**

1.  **Divide**: Split the unsorted list into two equal sub-lists.
2.  **Conquer**: Recursively sort each of these sub-lists using Merge Sort.
3.  **Combine/Merge**: Merge the two sorted sub-lists back into a single, sorted list. This merging step is key to Merge Sort's efficiency, as combining two already sorted lists is much faster than sorting a large list from scratch.

Merge Sort is well-regarded for its consistent performance (O(n log n) time complexity) and its stability (meaning it preserves the relative order of equal elements).

```python
def merge_sort(arr):
    """
    Sorts an array using the Merge Sort algorithm (Divide and Conquer).
    """
    # Base case: if the array has 0 or 1 element, it's already sorted
    if len(arr) <= 1:
        return arr

    # Divide: Find the middle point and divide the array into two halves
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    # Conquer: Recursively sort both halves
    left_half = merge_sort(left_half)       # Sort the left half
    right_half = merge_sort(right_half)     # Sort the right half

    # Combine: Merge the sorted halves
    return merge(left_half, right_half)

def merge(left, right):
    """
    Merges two sorted lists into a single sorted list.
    """
    merged_list = []
    left_idx, right_idx = 0, 0

    # Compare elements from both lists and append the smaller one
    while left_idx < len(left) and right_idx < len(right):
        if left[left_idx] < right[right_idx]:
            merged_list.append(left[left_idx])
            left_idx += 1
        else:
            merged_list.append(right[right_idx])
            right_idx += 1

    # Append any remaining elements from the left list (if any)
    while left_idx < len(left):
        merged_list.append(left[left_idx])
        left_idx += 1

    # Append any remaining elements from the right list (if any)
    while right_idx < len(right):
        merged_list.append(right[right_idx])
        right_idx += 1

    return merged_list

print("Merge Sort functions (merge_sort and merge) are defined.")
```

## Test Merge Sort Implementation

Here's an example of how to use the `merge_sort` function with an unsorted list.

```python
# Example usage:
unsorted_list = [38, 27, 43, 3, 9, 82, 10]
sorted_list = merge_sort(unsorted_list)

print(f"Unsorted List: {unsorted_list}")
print(f"Sorted List: {sorted_list}")

unsorted_list_2 = [5, 2, 8, 1, 9, 4]
sorted_list_2 = merge_sort(unsorted_list_2)

print(f"\nAnother Unsorted List: {unsorted_list_2}")
print(f"Another Sorted List: {sorted_list_2}")
```
