### Insertion Sort Explanation

Insertion sort is a simple sorting algorithm that builds the final sorted array (or list) one item at a time. It is much less efficient on large lists than more advanced algorithms such as quicksort, heapsort, or merge sort.

Here's how it works:
1.  **Iterate:** It iterates through the input array starting from the second element (assuming the first element is already sorted).
2.  **Key Selection:** For each element, it takes this element (the 'key') and compares it with the elements before it.
3.  **Shifting:** If the key is smaller than an element it's compared with, that larger element is shifted one position to the right.
4.  **Insertion:** This shifting continues until an element smaller than or equal to the key is found, or the beginning of the array is reached. The key is then inserted into its correct sorted position.

```python
def insertion_sort(arr):
    """
    Sorts an array using the Insertion Sort algorithm.
    """
    n = len(arr)

    # Traverse through 1 to len(arr)
    # The first element is considered sorted, so we start from the second.
    for i in range(1, n):
        key = arr[i]  # The element to be positioned in the sorted part

        # Initialize j to the index of the element just before the key
        j = i - 1

        # Move elements of arr[0...i-1], that are greater than key,
        # to one position ahead of their current position.
        # This creates a space for the 'key' to be inserted.
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j] # Shift element to the right
            j -= 1 # Move to the left to compare with the next element

        # Place the key at its correct position in the sorted part
        arr[j + 1] = key

    return arr

```

```python
# Example usage:
my_array = [12, 11, 13, 5, 6]
print("\n--- Demonstrating Insertion Sort ---")
sorted_arr = insertion_sort(my_array) # Call the simplified function
print(f"Initial Array: {[12, 11, 13, 5, 6]}") # Print initial array for clarity
print(f"Final Sorted Array: {sorted_arr}")
```
